## Conditional statements:
 conditional statements are questions that excute one seet of commands if its answer is true and another set if it is false

### examle:
* "if" : if it is raining take umbrella
* "else" : else do not take umbrella

```python
temp=35

if(temp>35):
    print("It is hot weather.")
else:
    print("It is not hot weather.") 
```

```python
temp=int(input("Enter the temperature:"))

if(temp>35):
    print("It is hot weather.")
else:
    print("It is not hot weather.") 
```

```python
score=3.2
if(score>=3.9):
    print("A+")
elif(score>=3.8):
    print("A")
elif(score>=3.6):
    print("B")
elif(score>=3.3):
    print("C")
else:
    print("Please focuse on study. You are not so good in studies.")
    
```

### Nasted if
an if statement placed inside another if statement

```python
age=int(input("Please enter your age: "))
if (age>=18):
    country=input("Please enter your country name:")
    if( country=="Pakistan"):
        print("Wow! Your are eligible for vote!!!")
    else:
        print("Sorry! you are not pakistani.")
else:
    print("You cannot vote.")
        
```

```python
# combine contition

age=int(input("Please enter your age: "))
country=input("Please enter your country name:").capitalize()

if (age>=18) and ( country=="Pakistan"):
    print("Wow! Your are eligible for vote!!!")
else:
    print("Sorry! you are not pakistani.")
    
```

## LOOPS

loops are used to iterate over sequences such as lists, strings, andother iterable data structures. it automatically stop after processing all items. used for searching through data

```python
for i in range(5):
    print("ALLAH")
```

```python
for char in "ALLAH":
    print(char)
```

```python
i=4
rows=i
col=i
matrix=[]
for r in range(rows):
    row=[]
    for c in range(col):
        row.append("ALLAH")
    matrix.append(row)

for row in matrix:
    print(" ".join(map(str, row)))
```

```python
for level in range(1, rows+1):
    print(" "*(rows-level)+"*" * (2*level-1))
```

### While loop:
 a while loop repeatedly executes a block of code as long as the specified condition remains true

 ### infinite loop
 a loop that never terminates because its condition always remains true

 #### for loop VS while loop
 * "for loop" known number of repetitions
 * counting to 10
 * "while loop" unknown depends on condition
 * waiting for input

```python
count=1
while count<=10:
    print(count)
    count= count+1
```

* output pairs like (0,0),(0,1),(1,0)....
