```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create 2x2 Subplots Grid
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# 1. Top-Left (0, 0): Line Plot
axes[0, 0].plot(x, y, color="blue")
axes[0, 0].set_title("Line Plot")

# 2. Top-Right (0, 1): Scatter Plot
axes[0, 1].scatter(x, y, color="red")
axes[0, 1].set_title("Scatter Plot")

# 3. Bottom-Left (1, 0): Histogram
np.random.seed(42)
hist_data = np.random.normal(0, 10, 200)
axes[1, 0].hist(hist_data, bins=20, color="green")
axes[1, 0].set_title("Histogram")

# 4. Bottom-Right (1, 1): Bar Chart
categories = ["A", "B", "C"]
values = [10, 20, 15]
axes[1, 1].bar(categories, values, color="magenta")
axes[1, 1].set_title("Bar Chart")

# Adjust spacing between plots
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Generate Sample Sales Dataset
np.random.seed(42)
n_rows = 5000

dates = pd.date_range(start="2022-01-01", periods=n_rows, freq="h")
df = pd.DataFrame(
    {
        "Order Date": dates,
        "Revenue": np.random.uniform(10, 500, size=n_rows),
        "Delivery Days": np.random.randint(1, 15, size=n_rows),
        "Customer Rating": np.random.uniform(1, 5, size=n_rows),
    }
)

# Resample Monthly Revenue
df_ts = df.set_index("Order Date")
monthly_revenue = df_ts["Revenue"].resample("ME").sum()

# Create 2x2 Subplots Grid
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# 1. Revenue Distribution (Histogram)
axes[0, 0].hist(df["Revenue"], bins=30, color="green")
axes[0, 0].set_title("Revenue Distribution")

# 2. Monthly Revenue Trend (Line Plot)
axes[0, 1].plot(
    monthly_revenue.index, monthly_revenue.values, color="red"
)
axes[0, 1].set_title("Monthly Revenue Trend")

# 3. Delivery Days Distribution (Histogram)
axes[1, 0].hist(df["Delivery Days"], bins=10, color="magenta")
axes[1, 0].set_title("Delivery Days Distribution")

# 4. Customer Rating Distribution (Histogram)
axes[1, 1].hist(df["Customer Rating"], bins=10, color="yellow")
axes[1, 1].set_title("Customer Rating Distribution")

plt.tight_layout()
plt.show()
```

```python
# 1. Sharing Y-Axis (sharey=True)
fig, axes = plt.subplots(1, 2, figsize=(8, 4), sharey=True)
axes[0].plot([1, 2, 3], [10, 20, 30], color="blue")
axes[0].set_title("Shared Y-Axis (Left)")

axes[1].plot([1, 2, 3], [15, 25, 35], color="orange")
axes[1].set_title("Shared Y-Axis (Right)")

plt.tight_layout()
plt.show()

# 2. Independent Y-Axes (sharey=False)
fig, axes = plt.subplots(1, 2, figsize=(8, 4), sharey=False)
axes[0].plot([1, 2, 3], [10, 20, 30], color="blue")
axes[0].set_title("Independent Y-Axis (Left)")

axes[1].plot([1, 2, 3], [100, 200, 300], color="orange")
axes[1].set_title("Independent Y-Axis (Right)")

plt.tight_layout()
plt.show()
```
