## OUTLIER:-
Outliers are data points that fall outside the normal range, representing extreme values that candistort the mean and median of the datset.

## Methods Of Outlier Identification And Removal:

* IQR Method
* z-score method
* domain-driven thresholds

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. Create Synthetic Dummy Data
np.random.seed(42)
n_samples = 5000

# Generate realistic base revenue data (centered around ~1000)
revenue_base = np.random.normal(loc=1000, scale=300, size=n_samples)

# Inject synthetic outliers (extremely high or low values)
n_outliers = 60
outlier_indices = np.random.choice(n_samples, size=n_outliers, replace=False)
revenue_base[outlier_indices] = np.random.uniform(
    3500, 8000, size=n_outliers
)  # Extreme values

# Construct the DataFrame
df = pd.DataFrame(
    {
        "Order ID": [f"ORD-{10000 + i}" for i in range(n_samples)],
        "Product Category": np.random.choice(
            ["Beauty", "Clothing", "Electronics", "Home"], size=n_samples
        ),
        "Revenue": np.round(revenue_base, 2),
    }
)

print(f"Original Dataset Shape: {df.shape}")
print(df.head())
print("\n" + "=" * 50 + "\n")


# =========================================================
# Method 1: IQR (Interquartile Range) Method
# =========================================================

# Calculate Q1, Q3, and IQR
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)
IQR = Q3 - Q1

# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
iqr_outliers = df[(df["Revenue"] < lower_bound) | (df["Revenue"] > upper_bound)]

print("=== 1. IQR Method Results ===")
print(f"Calculated Q1: {Q1:.2f}")
print(f"Calculated Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f} | Upper Bound: {upper_bound:.2f}")
print(f"Total Outliers Detected: {iqr_outliers.shape[0]}")
print("\nSample IQR Outliers Detected:")
print(iqr_outliers.head())

# Filter out outliers
df_iqr_cleaned = df[
    (df["Revenue"] >= lower_bound) & (df["Revenue"] <= upper_bound)
]
print(f"\nShape after IQR Removal: {df_iqr_cleaned.shape}")

print("\n" + "=" * 50 + "\n")


# =========================================================
# Method 2: Z-Score Method
# =========================================================

# Calculate absolute Z-scores
z_scores = np.abs(stats.zscore(df["Revenue"]))

# Detect outliers (Threshold > 3 Standard Deviations)
threshold = 3
z_outliers = df[z_scores > threshold]

print("=== 2. Z-Score Method Results ===")
print(f"Total Outliers Detected (Z-Score > {threshold}): {z_outliers.shape[0]}")
print("\nSample Z-Score Outliers Detected:")
print(z_outliers.head())

# Filter out outliers
df_zscore_cleaned = df[z_scores <= threshold]
print(f"\nShape after Z-Score Removal: {df_zscore_cleaned.shape}")
```
