## Numpy: 

Numpy provides a data structure that is efficient for array manipulation.

## Why Numpy:
* Pure python lists are slow and memory-inefficient.
* Numpy provides NDarray(N-dimensional array).
* Enables fast numerical operations(vectorized).
* Forms the foundation for scientific computing and ML libraries.

## Key Concept:
#### NDArray:
Core data structure:

#### Shape:

Dimensions of the array(rows * columns).

#### NDim:

Number of dimensons(1D,2D,3D,.....).

#### DType:

Data type of array elements (int,float,etc).

```python
## import numpy

import numpy as np
import time
```

```python
arr=np.array([[1,2,3],[4,5,6]])
print(arr.shape)
print(arr.ndim)
print(arr.dtype)
```

```python
py_list = list(range(1_000_000))
py_result=[]
start_time=time.time()
for i in py_list:
    py_result.append(i+5)

end_time=time.time()
print("python list time: ", end_time - start_time)
```

```python
np_array=np.arange(1_000_000)
start_time=time.time()
np_result=np_array + 5
end_time= time.time()
print("numpy array time: ", end_time - start_time)
```
