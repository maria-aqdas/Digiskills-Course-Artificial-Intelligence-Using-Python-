## Pandas Indexing and Selection - 'loc and  loci'

```python
import pandas as pd
```

```python
# sample data frame

data={
    'Name':['Maria', 'Alia', 'Ahmad', 'Abdullah'],
    'Age':[20,17,13,14],
    'City':['Kasur','Lahore','patoki','London'],
    'Salary':[1000000,600000,400000,900000]
}

df=pd.DataFrame(data)

print(" Original DataFrame:\n", df)
```

```python
# 1. Labal Based Selection ---(.loc)

subset_loc=df.loc[1:3,['Name','Salary']]
print("Subset with .loc:\n", subset_loc)
```

```python
# 2.Integer Based Selection (.iloc)

subset_iloc=df.iloc[0:2,1:3]
print("Subset with .iloc: \n",subset_iloc)
```

```python
# 3.Boolean Filtering:

high_salary=df[df['Salary']>500000]
print("High Salary Filter: \n", high_salary)
```

```python
# 4. Safe Assignment Using .loc

df.loc[df['Age']<15,'Salary']=500000
print("Low Age Salary Change: \n", df.loc[df['Age']<15,'Salary'])
```

## Import and Export - CSV,Excal and JSON

### Key Concepts: 
* Reading CSV: pd.read_csv()

#### Options:
* parse_dates=['column'] -> convert data columns
* usecols=[...] -> load specific columns
* nrows= 100 -> load subset to save memory
* Read Excal: pd.read_excal('file.xlsx',sheet_name='sheet1')
* Reading JSON: pd.read_json('file.json')

## Key Points:
* Use parse_dates when working with date/time data
* Exporting: csv,excal,json  ---> easy sharing and interoperability
* For large datasets, load only needed columns or rows

## String and Text Cleaning: 


```python
df=pd.DataFrame({
    'Category': ['  Electronics ', 'furniture', '  Home-appliances' ,  'electronics', 'FURNITURE']
})

print(df)
```

```python
# strip spaces and convert to lowercase
df['Category_clean']=df['Category'].str.strip().str.lower()
print(df)
```

```python
# replace hypens with space

df['Category_clean']=df['Category_clean'].str.replace('-', '')
print(df)
```
