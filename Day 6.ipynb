## Lambda Function

A lambda in python is a small anonymous function written in a single line

### when to use
* for simple, one-time task
* with built-in functions like map(), filter()

```python
add= lambda a, b: a+b
print(add(45,76))
```

```python
double_it = lambda n: n*2
print(double_it(5))
```

```python
nums = [1,2,3,4]
squares=list(map(lambda x:x*x, nums))
print(squares)
```

```python
numbers=[1,2,3,4,5,6,7,8]
odd_numbers=list(filter(lambda x: x%2 !=0, numbers))
print(odd_numbers)
```

## Class
 a class is a bluerint for creating objects. defines attributes (data) and methods (behavior).

```python
class car:
    def __init__(self, color):
        self.color=color

    def start(self):
        print(f"{self.color} car started")

my_car=car("red")
my_car.start()

```
