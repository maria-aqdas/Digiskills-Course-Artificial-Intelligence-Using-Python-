##  matplotlib

 

 matplotlib provides a figure that acts like a canvas on which various graphs and plots can be created

 ### axes
 the individual pot  area that contains the x-axis and y-axis
 

 ## 3d plotting
 represents data using three axes:
 * x-axis
 * y-axis
 * z-axis



## workflows
* stateful(pyplot style) functions operate on the current figure and axes(e.g., plt.plot())
* object- oriented(oo style) uses explicit figure and axes objects for greater control

## tip
* prefer the object-oriented(oo) style for reuseable and complex plots
* label the x-axis and y-axis
* use descriptive labels for plots

```python
import matplotlib.pyplot as plt

# 1. Create a figure and axis object using the subplots() function
fig, ax = plt.subplots()

# 2. Set the Title of the Plot
ax.set_title("Blank Figure Example")

# 3. Set Labels for X-axis and Y-axis
ax.set_xlabel("X-Axis Label")
ax.set_ylabel("Y-Axis Label")

# 4. Save the figure to disk with a specified DPI (Dots Per Inch)
fig.savefig("blank_plot.png", dpi=100)

# 5. Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create sample dummy data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create figure and axes
fig, ax = plt.subplots()

# Plot data
ax.plot(x, y, label="Sine Wave")

# Add Title and Axis Labels
ax.set_title("Matplotlib Object-Oriented Plot Example")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Amplitude")

# Save figure to disk
fig.savefig("sine_wave_plot.png", dpi=100)

# Display plot
plt.show()
```
