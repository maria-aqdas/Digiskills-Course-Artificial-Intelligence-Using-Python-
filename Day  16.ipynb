## Merging And Joining:

### Joining:
The function of joining is similar to how tables are combined in databases.

### Methods Of Joining:
* Inner Join
* Left Join
* Right Join
* Outer Join

```python
import pandas as pd
```

```python
# Sample Dataframe

customers=pd.DataFrame({
    'Customer ID': [1,2,3],
    'Names':['Alice', 'Bob','Charlie']
})

order=pd.DataFrame({
    'order ID':[101,102,103,104],
    'Customer ID': [1,2,2,3],
    'Amount': [250,150,200,300]
})


# Inner Merge on Customer ID:

df_inner=pd.merge(customers,order ,on='Customer ID',how='inner')
print(" INNER Merge: \n", df_inner)
```

```python
# Left Merge:

df_left=pd.merge(customers,order ,on='Customer ID',how='inner')
print(" Left Merge: \n", df_left)
```

```python
df_concat=pd.concat([customers,customers])
print(" INNER Merge: \n", df_inner)
```
