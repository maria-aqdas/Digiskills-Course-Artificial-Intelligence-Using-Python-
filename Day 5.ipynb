## Use of loops on lists

```python
for char in "Prophet Muhammad":
    print(char)
```

```python
count=0
for char in "Prophet Muhammad":
    if char=="m":
        count+=1
print(count)        
    
```

```python
s="python"
rev=""
for ch in s:
    rev=ch+rev

print(rev)    
```

```python
student={"Name":"Maria Aqdas", "Age":20,"CGPA": 3.86, "Course":"Data Science"}
for key in student:
    print(student[key])
```

```python
student={"Name":"Maria Aqdas", "Age":20,"CGPA": 3.86, "Course":"Data Science"}
for key in student:
    print(key,":",student[key])
```

# Defining and calling Functions

## Functio,

a function is a reuseable block of code that performs a spcific task.


### benefits of function

* increase reuse ability of code
* helps make programs organized

#### calling a function

after defining a function, we call it to execute.

```python
def greet():
    print("Assalam-o-Alikum!!")
greet()
```

```python
def greet(name):
    print("Assalam-o-Alikum!!",name,".")
greet("Maria Aqdas")
```

### Return values of function:

a function can also return a result after completing a specfic task; this returned result is called the return value of the function. 

```python
def add(a,b):
    return a+b
result= add(3,5)
print(result)
```

```python
def square(num):
    return num* num
result = square(5)
print(result)
```
