```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility

np.random.seed(42)

# Generate two normal distributions

group_a = np.random.normal(loc=70, scale=10, size=1000)
group_b = np.random.normal(loc=80, scale=12, size=1000)

# Plot overlapping histograms with density=True and alpha transparency

plt.hist(group_a, bins=30, density=True, alpha=0.6, label="Group A")
plt.hist(group_b, bins=30, density=True, alpha=0.6, label="Group B")

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Overlapping Histograms (Group A vs Group B)")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Generate sample dataset

np.random.seed(42)
n_rows = 5000

data = {
    "Unit Price": np.random.uniform(10, 300, size=n_rows),
    "Quantity": np.random.randint(1, 10, size=n_rows),
    "Discount": np.random.uniform(0.0, 0.3, size=n_rows),
}

df = pd.DataFrame(data)
df["Revenue"] = df["Unit Price"] * df["Quantity"]

# Take 1,000 random samples

df_sample = df.sample(n=1000, random_state=42)

# Standard Frequency Histogram (30 Bins)

fig, ax = plt.subplots(figsize=(12, 4))
ax.hist(df_sample["Revenue"], bins=30, edgecolor="black")

ax.set_xlabel("Revenue ($)")
ax.set_ylabel("Frequency")
ax.set_title("Revenue Distribution (Histogram)")
plt.show()
```

```python
# Density Histogram (Probability Density Function)
fig, ax = plt.subplots(figsize=(12, 4))
ax.hist(df_sample["Revenue"], bins=30, density=True, alpha=0.6, edgecolor="black")

ax.set_xlabel("Revenue ($)")
ax.set_ylabel("Density")
ax.set_title("Revenue Density Distribution")
plt.show()
```

```python
# Create Discounted Revenue column

df_sample["Discounted Revenue"] = df_sample["Revenue"] * (
    1 - df_sample["Discount"]
)

# Plot overlapping histograms comparing Original vs Discounted Revenue

fig, ax = plt.subplots(figsize=(12, 4))

ax.hist(
    df_sample["Revenue"],
    bins=30,
    alpha=0.6,
    label="Original Revenue",
)
ax.hist(
    df_sample["Discounted Revenue"],
    bins=30,
    alpha=0.6,
    label="Discounted Revenue",
)

ax.set_xlabel("Revenue ($)")
ax.set_ylabel("Frequency")
ax.set_title("Original Revenue vs Discounted Revenue Distribution")
ax.legend()
plt.show()
```
