```python
age=-20
if age < 0:
    raise ValueError("Age cannot be negative!!")
```

```python
class InValidAgeError(Exception):
    pass
```

```python
age=int(input("Enter age: "))
if age<0:
    raise InValidAgeError(f"Age cannot be negative! {age}")
```

```python
try:
    age= int(input("Enter Age: "))
    if age<0:
        raise InValidAgeError(f"Age Cannot be Negative!, {age}")

except InValidAgeError as e:
    print(e)
```

# data science

data science is a fundamental paradigm within AI in which data is studied, tanformed, and usd to  build predictive models

## data 

raw facts, figures or observations

## information

data that is transformed and made meaningful

## types of data

#### structured

rows/columns(databases,CSV)

#### unstructured

text, image, audio, video

#### semi-structured

JSON, XML

### importance of data in machine learning

* acts as the fule for machine learning algorithms
* directly impacts model accuracy and relability
* without quality data, even advanced models fail

### key characteristics of  high quality data

* relevance---> aligned with the problem
* accuracy---> error-free, reliable
* diversity---> covers fdifferent cases and variations
* sufficiency---> enough quantity for training

### role of data in ml pipeline

* collection -->gathering data
* pre-processing --> cleaning and transforming
* feature engineering --> extracting useful features
* model training --> feeding data into algorithms
* evaluation -->tsting model performance

## pandas vs numpy
