## Correlation And Covariance:

Measure the reletionshis between numerical variables

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
n_samples = 5000

quantity = np.random.randint(1, 10, size=n_samples)
unit_price = np.random.uniform(10.0, 500.0, size=n_samples).round(2)
discount = np.random.uniform(0.0, 0.3, size=n_samples).round(2)
delivery_days = np.random.randint(1, 14, size=n_samples)
```

```python
# Formula for Revenue with slight noise
revenue = (quantity * unit_price * (1 - discount)).round(2)

df = pd.DataFrame(
    {
        "Quantity": quantity,
        "Unit Price": unit_price,
        "Discount": discount,
        "Delivery Days": delivery_days,
        "Revenue": revenue,
    }
)

print(f"Dataset Shape: {df.shape}")
print(df.head())
```

```python
# Select only numerical columns for Covariance and Correlation Analysis
num_cols = df.select_dtypes(include=[np.number])
```

```python
# 1. Covariance Analysis (Absolute Units)
print("=== Covariance Matrix ===")
cov_matrix = num_cols.cov()
print(cov_matrix)
```

```python
# 2. Correlation Analysis (Scaled between -1 and +1)
print("\n=== Correlation Matrix ===")
corr_matrix = num_cols.corr()
print(corr_matrix)
```

```python
# 3. Sort Correlations for Revenue Column (Descending Order)
print("\n=== Revenue Correlations Sorted ===")
revenue_corr_sorted = corr_matrix["Revenue"].sort_values(ascending=False)
print(revenue_corr_sorted)
```

```python
# 4. Extract Minimum and Maximum Correlations for Revenue (Excluding self-correlation)
revenue_corr_no_self = revenue_corr_sorted.drop("Revenue")
min_corr = revenue_corr_no_self.min()
max_corr = revenue_corr_no_self.max()

print("\n=== Min and Max Correlation with Revenue ===")
print(f"Minimum Correlation: {min_corr:.4f}")
print(f"Maximum Correlation: {max_corr:.4f}")
```
