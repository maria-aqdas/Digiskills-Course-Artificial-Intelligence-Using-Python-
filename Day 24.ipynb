```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PART 1: Customizing & Exporting a Sine Plot


# 1. Generate data: linear space from 0 to 10 with 50 parameters
x = np.linspace(0, 10, 50)
y = np.sin(x)

# 2. Create a subplot of size 6x4
fig, ax = plt.subplots(figsize=(6, 4))

# 3. Plot the data with label, color, and linewidth
ax.plot(x, y, label='sin x', color='blue', linewidth=2)

# Customization A: Set x-ticks resolution (range 0 to 11 with step size 2)
ax.set_xticks(range(0, 12, 2))

# Customization B: Rotate x-axis labels by 30 degrees
ax.tick_params(axis='x', rotation=30)

# Customization C: Change legend location to upper right
ax.legend(loc='upper right')

# Set title and tight layout
ax.set_title("Sine Wave Customized Plot")
plt.tight_layout()

# 4. Export the figure to disk as a high-resolution PNG (300 DPI)
fig.savefig('sin_plot.png', dpi=300)
print("Saved 'sin_plot.png' successfully.")

plt.show()
```

```python

# PART 2: Bar Graph with Dummy Dataset

# 1. Create dummy data (Simulating the 5000 rows dataset mentioned in the video)
np.random.seed(42) # For reproducibility
data = {
    'Payment Method': np.random.choice(
        ['Card', 'Cash on Delivery', 'Wallet'], 
        size=5000, 
        p=[0.45, 0.40, 0.15] # Giving random weights to different payment types
    )
}
df = pd.DataFrame(data)

# 2. Count the occurrences of each payment method
counts = df['Payment Method'].value_counts()

# 3. Create subplot for the bar graph
fig2, ax2 = plt.subplots(figsize=(6, 4))

# 4. Create the bar graph using index and values
ax2.bar(counts.index, counts.values, color=['skyblue', 'orange', 'lightgreen'])

# Customization: Rotate x-axis labels by 45 degrees
ax2.tick_params(axis='x', rotation=45)
# Set labels and title
ax2.set_ylabel('Number of Orders')
ax2.set_title('Orders by Payment Method')

plt.tight_layout()
plt.show()
```
