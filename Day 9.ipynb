# exception handling:

an exception is an error that occurs during the execution of a program.

# importance of excaption handling

due to an error, python will stop the program and display an error message; instead, we should handle the error and allow the rogram to continue running.

## examples of errors

* dividing by zero
* file not found
* invalid user input

```python
num = int(input("Enter a number:"))
print(10/num)
```

```python
try:
    num = int(input("Enter a number:"))
    print(10/num)

except ZeroDivisionError:
    print("You can not divide by zero!!")
except ValueError:
    print("Thats not a number.")
except Exception as e:
    print(f" An Unexpected error occurred: {e}")
    
```
