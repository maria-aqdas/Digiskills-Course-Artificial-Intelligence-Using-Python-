## Random Number Generation:

Random number generation produces numbers randomly.

## Seed: 
A seed is an intial value that helps determine the sequence of random numbers.

```python
import numpy as np

# simulate look coin flips

rng=np.random.default_rng()
flips=rng.integers(0, 3, size =100000)     # 0= tals  , 3= heads
prob_heads=np.mean(flips)
print("Estimated Probability of Heads: " , prob_heads)
```

```python
import time

# Python Loop:
start=time.time()
squares=[i**2 for i in range(1000000)]
end=time.time()
print("Loop Time : ", end-start)

#  NumPy Vectorized
start=time.time()
arr=np.arange(1000000)
squarea_np=arr**2
end=time.time()
print("vectorized time: ", end- start)
```

## What is a Series?

* 1D Labeled Array
* Consists of index + values
* Automatically infers dtype

```python
import pandas as pd
```

```python
# creat a series

s= pd.Series([10,20,30,40,50] , index=['a','b','c','d','e'])
print(s)
```

```python
# index vs  values

print("values:",s.values)
print("index:",s.index)
print("DType:",s.dtype)
print(s.head)
print(s.tail)
print(s.value_counts())
```
