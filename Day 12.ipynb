```python
# import numpy

import numpy as np
```

```python
# print 1 dimentional array

arr= np.array([1,2,3])
print(arr)
```

```python
# print an array fill with zero 

print(np.zeros([3,3]))
```

```python
# print an array fill with ones 

print(np.ones([3,3]))
```

```python
# print an array fill with any number like 5 

print(np.full([3,3],5))
```

```python
# print an array that show even no b/w 1 to 50

print(np.arange(0,50,2))
```

```python
# print an array find 10 no. b/w 1 to 3 

print(np.linspace(0,3,10))
```

```python
# print dignol matrix 

print(np.eye(3))
```

```python
# print dignol matrix and we decide its dignol values

print(np.diag([1,2,3]))
```

```python
# rint an array those vales convert into float tye

np.array([1,2,3],dtype=float)
```

```python
# rint an array those vales convert into int tye

arr=np.array([1.8,2.6,9.9])
print(arr.astype(int))
```
