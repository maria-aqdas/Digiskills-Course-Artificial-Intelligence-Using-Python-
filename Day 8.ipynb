## inheritance

inheritance alows a new class to reuse the properties and methods of  an existing class.

```python
class animal:
    def speak(self):
        print("Some sound")

class dog(animal):
    pass

dog1=dog()
dog1.speak()
```

```python
class dog(animal):
    def bark(self):
        print("Woof!!")

childdog=dog()
childdog.bark()
childdog.speak()
```

## method overriding

```python
class animal:
    def speak(self):
        print("some sound")
class dog(animal):
    def speak(self):
        print("Woof!! woof!!")

dog1=dog()
dog1.speak()
```

## Polymorphism

polymorphism refers to the ability of a function or method to exist in multiple forms

```python
# method overloading

class calculater:
    def add(self, a, b=0, c=0):
        return a + b + c

calc1=calculater()
print(calc1.add(5))
print(calc1.add(5,34))
print(calc1.add(5,67,89))
```

```python
# polymorphism

class dog:
    def speak(self):
        print("woof!!")

class cat:
    def speak(self):
        print("Meow!!")

        
for pet in [dog(),cat()]:
    pet.speak()
```

```python
# method overriding

class vehical:
    def move(self):
        print("Vehicale is moving...........")

class car:
    def move(self):
        print("car is moving..........")

car1=car()
car1.move()

vehical1=vehical()
vehical1.move()
```

## special functions

 a special function defines built-in behavior automatically associated with certain operations

```python
class book:
    def __init__(self,title, author):
        self.title=title
        self.author=author

    def __str__(self):
        return f" '{self.title}' by {self.author}. He is a fomous novalist."

b= book("1984", "George Orwell")
print(b)
```

```python
class playlist:
    def __init__(self, songs):
        self.songs=songs
    def __len__(self):
        return len(self.songs)
    def __add__(self, other):
        return playlist(self.songs+ other.songs)

p1=playlist(["song1","song2"])
p2=playlist(["song3"])
print(len(p1))
print(len(p1+p2))
```

```python
class points:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __str__(self):
        return f"({self.x} , {self.y})"
    def __add__(self, other):
        if not isinstance(other,points):
            return NotImplemented
        return points(self.x + other.x, self.y+ other.y)

p1=points(2,4)
p2=points(4,7)
print(p1 +p2)
```
