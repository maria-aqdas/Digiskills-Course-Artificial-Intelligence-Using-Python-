```python
# import matplotlib

import matplotlib.pyplot as plt

# Sample Data:
days = [1, 2, 3, 4, 5, 6, 7]
sales = [150, 200, 180, 220, 300, 280, 350]

# Line Plot Setup
plt.plot(
    days,
    sales,
    marker="o",
    linewidth=2,
    alpha=0.8,
    label="Daily Sales",
)

# Axis Labels and Title
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Trend Over a Week")

# Add Legend and Display
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load or generate dataset (5,000 rows sample)
np.random.seed(42)
n_rows = 5000

dates = pd.date_range(start="2022-01-01", periods=n_rows, freq="h")
data = {
    "Order Date": dates,
    "Unit Price": np.random.uniform(10.0, 500.0, size=n_rows).round(2),
    "Quantity": np.random.randint(1, 10, size=n_rows),
}

df = pd.DataFrame(data)
df["Revenue"] = (df["Unit Price"] * df["Quantity"]).round(2)

# 1. Line Plot: Monthly Revenue Trend
df_ts = df.set_index("Order Date")
monthly_revenue = df_ts["Revenue"].resample("ME").sum()

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    linewidth=2,
)

ax.set_xlabel("Date")
ax.set_ylabel("Revenue ($)")
ax.set_title("Monthly Revenue Trend")
plt.show()
```

```python

# 2. Scatter Plot: Unit Price vs. Revenue (100 Random Samples)
# ---------------------------------------------------------
df_sample = df.sample(n=100, random_state=42)

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(
    df_sample["Unit Price"],
    df_sample["Revenue"],
    alpha=0.5,
)

ax.set_xlabel("Unit Price ($)")
ax.set_ylabel("Revenue ($)")
ax.set_title("Unit Price vs. Revenue")
plt.show()
```

```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)
n_rows = 5000

categories = ["Beauty", "Clothing", "Electronics", "Home"]
regions = ["East", "North", "South", "West"]

data = {
    "Product Category": np.random.choice(categories, size=n_rows),
    "Region": np.random.choice(regions, size=n_rows),
    "Revenue": np.random.uniform(50.0, 500.0, size=n_rows).round(2),
}

df = pd.DataFrame(data)
print(df.head())
```

```python
import matplotlib.pyplot as plt

# =========================================================
# 1. Simple Bar Chart: Total Revenue by Product Category
# =========================================================

# Group by Product Category and sum the Revenue
cat_revenue = df.groupby("Product Category")["Revenue"].sum()

fig, ax = plt.subplots(figsize=(7, 4))

# Plot bar chart
ax.bar(cat_revenue.index, cat_revenue.values)

# Set labels and title
ax.set_xlabel("Product Category")
ax.set_ylabel("Total Revenue ($)")
ax.set_title("Total Revenue by Product Category")

plt.show()
```

```python
# =========================================================
# 2. Grouped (Unstacked) Bar Chart: Revenue by Category & Region
# =========================================================

# Group by Product Category and Region, then unstack
cat_region_rev = df.groupby(["Product Category", "Region"])["Revenue"].sum().unstack()

# Plot grouped bar chart using Pandas wrapper on Matplotlib
ax = cat_region_rev.plot(kind="bar", figsize=(10, 4))

# Set labels and title
plt.xlabel("Product Category")
plt.ylabel("Revenue ($)")
plt.title("Revenue by Product Category across Regions")
plt.legend(title="Region")

plt.show()
```

```python
# =========================================================
# 3. Stacked Bar Chart: Revenue stacked by Region
# =========================================================

# Plot stacked bar chart
ax = cat_region_rev.plot(kind="bar", stacked=True, figsize=(10, 4))

# Set labels and title
plt.xlabel("Product Category")
plt.ylabel("Total Revenue ($)")
plt.title("Stacked Revenue by Product Category & Region")
plt.legend(title="Region")

plt.show()
```
