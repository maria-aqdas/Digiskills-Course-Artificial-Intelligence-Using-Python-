## What is EDA?

Exploratory Data Analysis (EDA) is the process of understanding data before modeling.

### Key Goals Of EDA:

* Identify Patterns, Trends, and Anomalies.
* Check data Quality.
* Understand feature Types.


## EDA MindSet and CheckList:
* Be Curious --> Ask Why and How
* Look for Relationship between Features.
* Document Assumptions and Findings.

## Functions For EDA Report:

* Data Dimensions (.shape)
* Column Types (.dtypes)
* Missing values (.isna().sum())
* Summary Statistics (.describe())
* Unique Values and Categories (.value_counts())

### Example DataSet:
#### Customer Transactions:

* How many missing values are present in each column?
* What s the distribution of product categories?
* Which products are purchased most frequently?
* Are there any outliers in the transaction amounts?
* How to purchase amounts vary by customer region?


### EXAMPLE:
#### 5 EDA Questions:

* Is there any missing data? ---> Quantity has *  Missing values

### Key TakeAways: 

* Always define goals before starting EDA.
* Identify target and features types early.
* use EDA questions to guide analysis .
